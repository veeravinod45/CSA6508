```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen2.5-0.5B-Instruct"

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading model...")

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
).to(device)

print("Model loaded successfully!")
print("Running on:", device)
```

```python
def generate_response(prompt):

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(text, return_tensors="pt").to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=True,
            temperature=0.7,
            top_p=0.9
        )

    response = outputs[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(response, skip_special_tokens=True)


question = "Explain Artificial Intelligence and its applications in engineering."

answer = generate_response(question)

print("User Prompt:")
print(question)

print("\nLocal LLM Response:")
print(answer)
```

```python
from reportlab.pdfgen import canvas

pdf_file = "Engineering_Notes.pdf"

c = canvas.Canvas(pdf_file)

content = [
    "ENGINEERING STUDY NOTES",
    "",
    "Topic: Artificial Intelligence and Machine Learning",
    "",
    "Artificial Intelligence (AI) is a branch of computer science",
    "that enables machines to perform tasks requiring human intelligence.",
    "",
    "Machine Learning is a subset of AI that allows computers",
    "to learn from data without being explicitly programmed.",
    "",
    "Deep Learning uses neural networks with multiple layers",
    "to process complex data such as images and speech.",
    "",
    "Natural Language Processing enables computers to understand",
    "and process human languages.",
    "",
    "Applications of AI include healthcare, robotics, education,",
    "transportation, manufacturing and engineering.",
    "",
    "Python is widely used for AI development because of its",
    "libraries such as NumPy, Pandas, TensorFlow and PyTorch."
]

y = 800

for line in content:
    c.drawString(50, y, line)
    y -= 30

c.save()

print("Engineering PDF created successfully!")
print("File name:", pdf_file)
```

```python
from pypdf import PdfReader

reader = PdfReader("Engineering_Notes.pdf")

document_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        document_text += text + "\n"

print("PDF text extracted successfully!")
print("\nExtracted Document Content:\n")
print(document_text)
```

```python
def split_text(text, chunk_size=100, overlap=20):

    words = text.split()
    chunks = []

    for i in range(0, len(words), chunk_size - overlap):
        chunk = " ".join(words[i:i + chunk_size])

        if chunk:
            chunks.append(chunk)

    return chunks


chunks = split_text(document_text)

print("Document split successfully!")
print("Total chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}:")
    print(chunk)
```

```python
import faiss
import numpy as np

embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True
)

embeddings = np.array(embeddings, dtype="float32")

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("FAISS Vector Database created successfully!")
print("Total vectors stored:", index.ntotal)
```

```python
def retrieve_context(question, top_k=2):

    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    )

    question_embedding = np.array(
        question_embedding,
        dtype="float32"
    )

    distances, indices = index.search(
        question_embedding,
        min(top_k, len(chunks))
    )

    results = []

    for i in indices[0]:
        if i >= 0:
            results.append(chunks[i])

    return "\n".join(results)


question = "What is Machine Learning?"

context = retrieve_context(question)

print("User Question:", question)
print("\nRetrieved Context:")
print(context)
```

```python
from groq import Groq
import os

client = Groq(api_key=os.environ["GROQ_API_KEY"])

def rag_answer(question):

    context = retrieve_context(question)

    prompt = f"""
    You are an engineering study assistant.

    Answer the question using only the provided document context.

    If the answer is not available in the context,
    clearly say that the document does not contain the answer.

    Document Context:
    {context}

    Student Question:
    {question}

    Provide a simple and clear answer.
    """

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_completion_tokens=300
    )

    return response.choices[0].message.content


question = "What is Machine Learning?"

answer = rag_answer(question)

print("Question:", question)
print("\nRAG Generated Answer:")
print(answer)
```

```python
import gradio as gr

def engineering_assistant(question):
    if not question.strip():
        return "Please enter a question."

    return rag_answer(question)

demo = gr.Interface(
    fn=engineering_assistant,
    inputs=gr.Textbox(
        label="Ask an Engineering Question",
        placeholder="Example: What is Deep Learning?",
        lines=3
    ),
    outputs=gr.Textbox(
        label="AI Generated Answer",
        lines=6
    ),
    title="Engineering AI Assistant",
    description="Ask questions based on your uploaded engineering notes using RAG and an LLM."
)

demo.launch(share=True)
```
