```python

# Experiment 6: Zero-shot, One-shot, Few-shot

def zero_shot(text):
    positive = ["good", "great", "excellent", "happy",
                "love", "amazing"]
    negative = ["bad", "poor", "worst", "hate", "terrible"]

    text = text.lower()

    if any(word in text for word in positive):
        return "Positive"
    elif any(word in text for word in negative):
        return "Negative"
    else:
        return "Neutral"


def one_shot(text):
    # Example: "I love this product" -> Positive
    return zero_shot(text)


def few_shot(text):
    # Multiple examples are represented by word lists
    positive = ["good", "great", "excellent", "happy",
                "love", "amazing"]
    negative = ["bad", "poor", "worst", "hate", "terrible"]

    text = text.lower()

    if any(word in text for word in positive):
        return "Positive"
    elif any(word in text for word in negative):
        return "Negative"
    else:
        return "Neutral"


review = input("Enter a review: ")

print("Zero-shot result:", zero_shot(review))
print("One-shot result:", one_shot(review))
print("Few-shot result:", few_shot(review))
```

```python

from google import genai
import getpass

# Enter your API key only if GEMINI_API_KEY is not already defined
try:
    GEMINI_API_KEY
except NameError:
    GEMINI_API_KEY = getpass.getpass("Enter your Gemini API key: ")

client = genai.Client(api_key=GEMINI_API_KEY)

# Structured prompt for Python program
python_prompt = """
Task: Write a Python program.
Problem: Check whether a number is even or odd.
Input: One integer from the user.
Output: Print Even or Odd.
Rules:
1. Use simple Python code for beginners.
2. Use the input() function.
3. Return only the Python program.
"""

# Structured prompt for SQL query
sql_prompt = """
Task: Write an SQL query.
Table name: Students
Columns: id, name, marks
Requirement: Display students whose marks are greater than 80.
Rules:
1. Use a SELECT statement.
2. Return only the SQL query.
"""

model_name = "gemini-flash-lite-latest"

try:
    # Generate Python program
    python_response = client.models.generate_content(
        model=model_name,
        contents=python_prompt
    )

    print("GENERATED PYTHON PROGRAM:")
    print(python_response.text)

    # Generate SQL query
    sql_response = client.models.generate_content(
        model=model_name,
        contents=sql_prompt
    )

    print("\nGENERATED SQL QUERY:")
    print(sql_response.text)

except Exception as e:
    print("Error:", e)
```

```python

from google import genai
import getpass

try:
    GEMINI_API_KEY
except NameError:
    GEMINI_API_KEY = getpass.getpass("Enter your Gemini API key: ")

client = genai.Client(api_key=GEMINI_API_KEY)
model_name = "gemini-flash-lite-latest"

# Zero-shot prompting: no examples
zero_shot = """
Classify the sentiment of this review as Positive or Negative.
Review: The movie was excellent and enjoyable.
Return only the sentiment.
"""

# One-shot prompting: one example
one_shot = """
Classify the sentiment as Positive or Negative.

Example:
Review: The food was delicious.
Sentiment: Positive

Review: The movie was excellent and enjoyable.
Sentiment:
"""

# Few-shot prompting: multiple examples
few_shot = """
Classify the sentiment as Positive or Negative.

Example 1:
Review: The product is wonderful.
Sentiment: Positive

Example 2:
Review: The service was terrible.
Sentiment: Negative

Example 3:
Review: I really loved this movie.
Sentiment: Positive

Review: The movie was excellent and enjoyable.
Sentiment:
"""

prompts = {
    "Zero-shot": zero_shot,
    "One-shot": one_shot,
    "Few-shot": few_shot
}

try:
    for strategy, prompt in prompts.items():
        response = client.models.generate_content(
            model=model_name,
            contents=prompt
        )

        print("\n" + "=" * 35)
        print(strategy + " Prompting")
        print("=" * 35)
        print(response.text)

except Exception as e:
    print("Error:", e)
```
