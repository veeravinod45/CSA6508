EXP 3


```python
from transformers import AutoTokenizer

bert = AutoTokenizer.from_pretrained("bert-base-uncased")
gpt = AutoTokenizer.from_pretrained("gpt2")

text = "I love learning Python"

bert_tokens = bert.tokenize(text)
bert_ids = bert.convert_tokens_to_ids(bert_tokens)

gpt_tokens = gpt.tokenize(text)
gpt_ids = gpt.convert_tokens_to_ids(gpt_tokens)

print("Input Text:", text)
print("BERT Tokens:", bert_tokens)
print("BERT Token IDs:", bert_ids)
print("GPT Tokens:", gpt_tokens)
print("GPT Token IDs:", gpt_ids)
```

EXP 4

```python

import torch
from transformers import AutoTokenizer, AutoModel

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

sentences = [
    "We sat on the river bank",
    "I deposited money in the bank"
]

for sentence in sentences:
    inputs = tokenizer(sentence, return_tensors="pt")

    with torch.no_grad():
        outputs = model(**inputs)

    embeddings = outputs.last_hidden_state

    tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

    print("\nSentence:", sentence)
    print("Tokens:", tokens)
    print("Embedding shape:", embeddings.shape)
    print("Bank embedding:", embeddings[0, tokens.index("bank"), :5])
```

EXP 8

```python
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained(
    "google-bert/bert-base-uncased"
)

sentence = "I love learning Natural Language Processing."

tokens = tokenizer.tokenize(sentence)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Original sentence:", sentence)
print("Tokens:", tokens)
print("Token IDs:", token_ids)

encoded = tokenizer(
    sentence,
    add_special_tokens=True
)

print("Tokens with special tokens:",
      tokenizer.convert_ids_to_tokens(encoded["input_ids"]))
print("Input IDs:", encoded["input_ids"])
```

EXP 9

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

model_name = "openai-community/gpt2"

gpt_tokenizer = GPT2Tokenizer.from_pretrained(model_name)
gpt_model = GPT2LMHeadModel.from_pretrained(model_name)

prompt = "Artificial intelligence is"

inputs = gpt_tokenizer(prompt, return_tensors="pt")

outputs = gpt_model.generate(
    **inputs,
    max_new_tokens=40,
    do_sample=True,
    temperature=0.8,
    top_k=50,
    pad_token_id=gpt_tokenizer.eos_token_id
)

generated_text = gpt_tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Prompt:", prompt)
print("Generated text:", generated_text)
```
